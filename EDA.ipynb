```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Load the cleaned dataset
df = pd.read_csv('cleaned_titanic.csv')

# Display basic information
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Statistical Summary of Numerical Features
print("=== Numerical Summary ===")
display(df.describe().T)

# Survival Rate Breakdown
print("\n=== Survival Rate by Gender ===")
display(df.groupby('Sex')['Survived'].agg(['count', 'mean']))

print("\n=== Survival Rate by Passenger Class ===")
display(df.groupby('Pclass')['Survived'].agg(['count', 'mean']))
```

## Summary Statistics Overview

- **Overall Survival Rate:** 38.38% (342 out of 891 passengers survived).
- **Average Age:** ~29.1 years (ranging from infants under 1 year to 80 years old).
- **Average Fare:** $32.20 (median fare is $14.45, indicating right-skewed high values).
- **Gender Distribution:** Female passengers had a ~74.2% survival rate compared to ~18.9% for male passengers.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Survival Rate by Class and Sex
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', palette='Set2', ax=axes[0, 0])
axes[0, 0].set_title('1. Survival Rate by Class & Sex', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Survival Rate')

# Plot 2: Age Distribution by Survival
sns.kdeplot(data=df, x='Age', hue='Survived', common_norm=False, fill=True, palette='coolwarm', ax=axes[0, 1])
axes[0, 1].set_title('2. Age Distribution Density by Survival Status', fontsize=12, fontweight='bold')

# Plot 3: Survival Rate by Family Size
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
sns.barplot(data=df, x='FamilySize', y='Survived', palette='Blues_d', ax=axes[1, 0])
axes[1, 0].set_title('3. Survival Rate by Family Size', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Survival Rate')

# Plot 4: Log Fare Distribution by Survival
sns.boxplot(data=df, x='Survived', y='Fare', palette='Set3', ax=axes[1, 1])
axes[1, 1].set_yscale('log')
axes[1, 1].set_title('4. Fare Distribution (Log Scale)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

## Key Exploratory Insights & Findings

1. **Gender Priority ("Women First"):**
   Female passengers survived at a significantly higher rate (**74.20%**) than male passengers (**18.89%**).
2. **Socio-Economic Privilege:**
   Class status strongly dictated survival probability. First-class passengers achieved a **62.96%** survival rate, compared to **24.24%** for third-class passengers.
3. **Child Evacuation Trend:**
   Children aged 12 and under experienced higher survival rates (**57.97%**) relative to working-age adults.
4. **Family Size Dynamics:**
   Solo travelers had lower survival rates (**30.35%**). Small families (2–4 members) fared best (peaking at **72.41%** for families of 4), while large families (5+ members) faced steep drops in survival (< 20%).
5. **Fare Anomalies & High Variance:**
   The fare column spans from **$0.00** (15 complimentary/crew records) to extreme outliers of **$512.33**, demonstrating significant economic disparity aboard the vessel.
